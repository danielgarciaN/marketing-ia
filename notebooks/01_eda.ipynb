# 01 - EDA

Exploratory analysis for revenue, countries, products, orders and customer behavior.

```python
from pathlib import Path
import pandas as pd

root = Path.cwd().parents[0]
data_path = root / 'backend' / 'app' / 'data' / 'processed' / 'online_retail_clean.csv'
df = pd.read_csv(data_path, parse_dates=['InvoiceDate'])
df.head()
```

```python
summary = {
    'rows': len(df),
    'customers': df['CustomerID'].nunique(),
    'orders': df['InvoiceNo'].nunique(),
    'revenue': df['Revenue'].sum(),
    'countries': df['Country'].nunique(),
}
summary
```
