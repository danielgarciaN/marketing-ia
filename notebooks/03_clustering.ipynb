# 03 - Clustering

KMeans evaluation, cluster interpretation and PCA projection.

```python
from pathlib import Path
import json
import pandas as pd

root = Path.cwd().parents[0]
processed = root / 'backend' / 'app' / 'data' / 'processed'
rfm = pd.read_csv(processed / 'customers_rfm.csv')
metrics = json.loads((processed / 'model_metrics.json').read_text())
metrics
```

```python
rfm.groupby('Cluster').agg(
    customers=('CustomerID', 'count'),
    avg_recency=('Recency', 'mean'),
    avg_frequency=('Frequency', 'mean'),
    avg_monetary=('Monetary', 'mean')
).round(2)
```
